```python
#loading libraries:
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv(r"C:\Users\Anushkha Bhagra\Downloads\ML\classification\Churn_Modelling.xls")
df.head()
```

```python
# drop the useless columns:
df.drop(columns=["RowNumber","CustomerId","Surname"],inplace=True)
```

```python
#handling categorical data:
label = LabelEncoder()
df["encoded_geo"] = label.fit_transform(df["Geography"])
var1 = dict(zip(df["Geography"],df["encoded_geo"]))
print(var1)

df["encoded_gender"] = label.fit_transform(df["Gender"])
var2 = dict(zip(df["Gender"],df["encoded_gender"]))
print(var2)
```

```python
#dropping columns:
df.drop(columns=["Geography", "Gender"],inplace=True)
```

```python
df.head()
```

```python
#featuring:
x = df.drop(columns=["Exited"])
y = df["Exited"]
```

```python
#data splittinf:
x_train,x_test,y_train,y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
scalar = StandardScaler()
x_train_scaler = scalar.fit_transform(x_train)
x_test_scaler = scalar.transform(x_test)
```

```python
# model selection:
logistic = LogisticRegression(max_iter= 10000)

knn = KNeighborsClassifier(n_neighbors=5)
```

```python
#model training using logistic regression:
logistic.fit(x_train_scaler,y_train)
```

```python
#model testing:
y_pred = logistic.predict(x_test_scaler)
accuracy_score(y_test,y_pred)
```

```python
#accuracy on train data:
y_pred_train = logistic.predict(x_train_scaler)
accuracy_score(y_train,y_pred_train)
```

```python
# model training using knn:
knn.fit(x_train_scaler,y_train)
```

```python
#model testing:
y_pred = knn.predict(x_test_scaler)
accuracy_score(y_test,y_pred)
```

```python
#accuracy on train data:
y_pred_train = knn.predict(x_train_scaler)
accuracy_score(y_train,y_pred_train)
```
