```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import RandomizedSearchCV
```

```python
df = pd.read_csv(r"C:\Users\Anushkha Bhagra\Downloads\ML\ensemble_learning\archive (1)\NFLX.csv")
df.head()
```

```python
df.drop(columns=["Adj Close","Date"],inplace=True)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
#feature selection:
x = df.drop(columns=["Close"])
y = df["Close"]
```

```python
#data selection:
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#scaling:
scaler = StandardScaler()
x_train_scalar = scaler.fit_transform(x_train)
x_test_scaler = scaler.transform(x_test)
```

```python
para_grid = {"max_depth": [3,5,6,7,9,11],
            "n_estimators": [100,150,200,250,300]}
```

```python
tuning_model = RandomForestRegressor(random_state=42)

random_model = RandomizedSearchCV(param_distributions = para_grid,
                                 estimator = tuning_model,
                                 n_jobs= -1,
                                 cv =5,
                                 n_iter = 30,
                                 scoring= "r2",
                                 verbose=1)
random_model.fit(x_train_scalar,y_train)
```

```python
y_pred = random_model.predict(x_test_scaler)
r2_score(y_test,y_pred)
```

```python
y_pred_train = random_model.predict(x_train_scalar)
r2_score(y_train,y_pred_train)
```
