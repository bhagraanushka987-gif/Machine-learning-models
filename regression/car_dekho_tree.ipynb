```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.tree import DecisionTreeRegressor # decision tree
from sklearn.ensemble import RandomForestRegressor # random forest
```

```python
df = pd.read_csv(r"C:\Users\Anushkha Bhagra\Downloads\ML\cardekho.csv")
```

```python
df.head()
```

```python
#preprocessing of data:
df.isnull().sum()
```

```python
df["mileage"] = df["mileage"].fillna(df["mileage"].mean())
df["engine"] = df["engine"].fillna(df["engine"].mean())
df["seats"] = df["seats"].fillna(df["seats"].mean())
```

```python
df.drop_duplicates(inplace=True)
```

```python
# encoding of categorical data:
label = LabelEncoder()
df["encoded_fuel"] = label.fit_transform(df["fuel"])
var1 = dict(zip(df["fuel"],df["encoded_fuel"]))
print(var1)

df["encoded_seller"] = label.fit_transform(df["seller_type"])
var2 = dict(zip(df["seller_type"],df["encoded_seller"]))
print(var2)

df["encoded_trans"] = label.fit_transform(df["transmission"])
var3 = dict(zip(df["transmission"],df["encoded_trans"]))
print(var3)

df["encoded_owner"] = label.fit_transform(df["owner"])
var4 = dict(zip(df["owner"],df["encoded_owner"]))
print(var4)
```

```python
#dropping the duplicate columns:
df.drop(columns=["fuel","seller_type","transmission","owner"],inplace=True)
```

```python
df.head()
```

```python
#feature selection :
x = df.drop(columns=["selling_price"])
y = df["selling_price"]
```

```python
#data splitting:
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#scaling:
#scalar = StandardScaler()
#x_train_scalar = scalar.fit_transform(x_train)
#x_test_scalar = scalar.fit_transform(x_test)
```

```python
#model selection:
tree = DecisionTreeRegressor(max_depth=7,random_state=42)
random = RandomForestRegressor(n_estimators=100,max_depth= 5, random_state=42)
```

```python
#model training using rndom tree
tree.fit(x_train,y_train)
```

```python
#accuracy on test data:
y_pred = tree.predict(x_test)
r2_score(y_test,y_pred)
```

```python
#model training using random forest:
random.fit(x_train,y_train)
```

```python
#model accuracy:
#accuracy on test data:
y_pred = random.predict(x_test)
r2_score(y_test,y_pred)
```
