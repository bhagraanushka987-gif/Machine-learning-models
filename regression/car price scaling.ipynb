```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import r2_score 
```

```python
df= pd.read_csv(r"C:\Users\Anushkha Bhagra\Downloads\ML\cardekho.csv")
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
#filling the null values
df["mileage"] = df["mileage"].fillna(df["mileage"].mean())
df["engine"] = df["engine"].fillna(df["engine"].mean())
df["seats"] = df["seats"].fillna(df["seats"].mean())
```

```python
#dropping the duplicates:
df.drop_duplicates(inplace=True)
```

```python
df["fuel"].value_counts()
```

```python
df.seller_type.value_counts()
```

```python
df.transmission.value_counts()
```

```python
df.owner.value_counts()
```

```python
#handling categorical data:
df["fuel"] = df["fuel"].map({"Diesel":0,"Petrol":1,"CNG":2,"LPG":3})
df["seller_type"] = df["seller_type"].map({"Individual":0,"Dealer":1,"Trustmark Dealer":2})
df["transmission"] = df["transmission"].map({"Manual":0,"Automatic":1})
df["owner"] = df["owner"].map({"First Owner":0,"Second Owner":1,"Third Owner":2,"Fourth & Above Owner":3,"Test Drive Car":4})
```

```python
df.head()
```

```python
#feature selection:
x = df.drop(columns=["selling_price"])
y = df["selling_price"]
```

```python
#data splitting:
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

```python
#data scaling: only of the input data means only x_train and x_test
scalar = StandardScaler()
x_train_scale = scalar.fit_transform(x_train)
x_test_scale = scalar.fit_transform(x_test)
```

```python
#model selection:
model = LinearRegression()

#model training:
model.fit(x_train_scale,y_train)
```

```python
#accuracy:
y_pred = model.predict(x_test_scale)
r2_score(y_test,y_pred)
```
