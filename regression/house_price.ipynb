```python
import pandas as pd
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv(r"C:\Users\Anushkha Bhagra\Downloads\ML\House_price_dataset.csv")
df.head()
```

```python
#preprocessing of the data:
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
label = LabelEncoder()
df["encoded_street"] = label.fit_transform(df["street"])
var1 = dict(zip(df["street"],df["encoded_street"]))
print(var1)

df["encoded_city"] = label.fit_transform(df["city"])
var2 = dict(zip(df["city"],df["encoded_city"]))
print(var2)

df["encoded_zip"] = label.fit_transform(df["statezip"])
var3 = dict(zip(df["statezip"],df["encoded_zip"]))
print(var3)

df["encoded_country"] = label.fit_transform(df["country"])
var4 = dict(zip(df["country"],df["encoded_country"]))
print(var4)
```

```python
df.drop(columns=["date","street","city","statezip","country"],inplace=True)
```

```python
df.head()
```

```python
#feature selection:
x = df.drop(columns=["price"])
y = df["price"]
```

```python
#data splitting
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
#model selection 
model = LinearRegression()
tree = DecisionTreeRegressor(max_depth=7,random_state=42)
random = RandomForestRegressor(n_estimators=100,max_depth=7,random_state=42)
```

```python
#model training using linear regression:
model.fit(x_train,y_train)
```

```python
#model training using desicion tree regression
tree.fit(x_train,y_train)
```

```python
#model training using random forest regression
random.fit(x_train,y_train)
```

```python
y_pred = model.predict(x_test)
r2_score(y_test,y_pred)
```

```python
y_pred = tree.predict(x_test)
r2_score(y_test,y_pred)
```

```python
y_pred = random.predict(x_test)
r2_score(y_test,y_pred)
```
