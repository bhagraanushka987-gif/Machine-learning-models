```python
import pandas as pd 
from sklearn.model_selection import train_test_split
from sklearn.model_selection import GridSearchCV, RandomizedSearchCV
from sklearn.metrics import r2_score
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.ensemble import RandomForestRegressor
```

```python
df = pd.read_csv(r"C:\Users\Anushkha Bhagra\Downloads\ML\regression\StudentPerformanceFactors.csv")
df.head()
```

```python
df.isnull().sum()
```

```python
df["Teacher_Quality"] = df["Teacher_Quality"].fillna(df["Teacher_Quality"].mode()[0])
df["Parental_Education_Level"] = df["Parental_Education_Level"].fillna(df["Parental_Education_Level"].mode()[0])
df["Distance_from_Home"] = df["Distance_from_Home"].fillna(df["Distance_from_Home"].mode()[0])
```

```python
df.drop_duplicates(inplace=True)
```

```python
#encoding:
label = LabelEncoder()
df["encoded_parent"] = label.fit_transform(df["Parental_Involvement"]) 
var1 = dict(zip(df["Parental_Involvement"],df["encoded_parent"]))
print(var1)

df["encoded_resource"] = label.fit_transform(df["Access_to_Resources"]) 
var2 = dict(zip(df["Access_to_Resources"],df["encoded_resource"]))
print(var2)

df["encoded_activities"] = label.fit_transform(df["Extracurricular_Activities"]) 
var3 = dict(zip(df["Extracurricular_Activities"],df["encoded_activities"]))
print(var3)

df["encoded_learning"] = label.fit_transform(df["Learning_Disabilities"]) 
var4 = dict(zip(df["Learning_Disabilities"],df["encoded_learning"]))
print(var4)

df["encoded_edu"] = label.fit_transform(df["Parental_Education_Level"]) 
var5 = dict(zip(df["Parental_Education_Level"],df["encoded_edu"]))
print(var5)

df["encoded_motivation"] = label.fit_transform(df["Motivation_Level"]) 
var6 = dict(zip(df["Motivation_Level"],df["encoded_motivation"]))
print(var6)

df["encoded_internet"] = label.fit_transform(df["Internet_Access"]) 
var7 = dict(zip(df["Internet_Access"],df["encoded_internet"]))
print(var7)

df["encoded_home"] = label.fit_transform(df["Distance_from_Home"]) 
var8 = dict(zip(df["Distance_from_Home"],df["encoded_home"]))
print(var8)


df["encoded_income"] = label.fit_transform(df["Family_Income"]) 
var9 = dict(zip(df["Family_Income"],df["encoded_income"]))
print(var9)


df["encoded_teacher"] = label.fit_transform(df["Teacher_Quality"]) 
var10 = dict(zip(df["Teacher_Quality"],df["encoded_teacher"]))
print(var10)

df["encoded_school"] = label.fit_transform(df["School_Type"]) 
var11 = dict(zip(df["School_Type"],df["encoded_school"]))
print(var11)

df["encoded_peer"] = label.fit_transform(df["Peer_Influence"]) 
var12 = dict(zip(df["Peer_Influence"],df["encoded_peer"]))
print(var12)

df["encoded_gender"] = label.fit_transform(df["Gender"]) 
var13 = dict(zip(df["Gender"],df["encoded_gender"]))
print(var13)
```

```python
df.drop(columns=["Parental_Involvement","Access_to_Resources","Extracurricular_Activities","Motivation_Level","Internet_Access","Family_Income","Teacher_Quality","School_Type","Peer_Influence","Learning_Disabilities","Parental_Education_Level"	,"Distance_from_Home","Gender"],inplace=True)
```

```python
df.head()
```

```python
#feature selection:
x = df.drop(columns=["Exam_Score"])
y = df["Exam_Score"]
```

```python
#data splitting:
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
#hyper parameter:
para_grid = {"max_depth": [3,5,7,8,9,11],
            "n_estimators": [100,150,200,250,300,350]}
```

```python
#tunning:
tunning = RandomForestRegressor(random_state=42)
```

```python
grid_model = GridSearchCV(param_grid= para_grid,
                          cv = 5,
                          n_jobs=-1,
                          scoring = "r2",
                          estimator = tunning)
grid_model.fit(x_train,y_train)
```

```python
grid_model.best_params_
```

```python
model = grid_model.best_estimator_
```

```python
y_pred = model.predict(x_test)
r2_score(y_test,y_pred)
```

```python
y_pred_train = model.predict(x_train)
r2_score(y_train,y_pred_train)
```
